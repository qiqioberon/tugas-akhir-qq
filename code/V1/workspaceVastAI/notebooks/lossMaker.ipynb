```python
# =========================================
# CELL 1 — Setup dan cari file history.csv
# =========================================

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Ganti path ini jika folder output kamu berbeda
FINETUNE_ROOT = Path.cwd().parent / "outputs" 

# Cari semua file history.csv dari seluruh run
history_files = sorted(FINETUNE_ROOT.glob("**/history.csv"))

print(f"Jumlah history.csv yang ditemukan: {len(history_files)}")
for i, hf in enumerate(history_files, 1):
    print(f"{i}. {hf}")
```

```python
# =========================================
# CELL 2 — Baca salah satu history.csv
# =========================================

# Pilih file history ke berapa dari daftar sebelumnya
selected_idx = 0   # 0 = file pertama, 1 = file kedua, dst

history_path = history_files[selected_idx]
print("File yang dipilih:")
print(history_path)

df_hist = pd.read_csv(history_path)

print("\nKolom yang tersedia:")
print(df_hist.columns.tolist())

print("\nPreview isi history:")
display(df_hist.head())
```

```python
# =========================================
# CELL 3 — Save grafik Train MAE vs Val MAE
#          untuk semua history.csv
# =========================================

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Folder untuk menyimpan semua grafik
save_dir = Path.cwd() / "mae_plots"
save_dir.mkdir(parents=True, exist_ok=True)

saved_files = []

for history_path in history_files:
    df_hist = pd.read_csv(history_path)

    # Skip kalau kolom wajib tidak ada
    required_cols = {"epoch", "train_mae", "mae_mean"}
    if not required_cols.issubset(df_hist.columns):
        print(f"[SKIP] Kolom tidak lengkap: {history_path}")
        continue

    # Cari epoch terbaik berdasarkan validation MAE minimum
    best_idx = df_hist["mae_mean"].idxmin()
    best_epoch = df_hist.loc[best_idx, "epoch"]
    best_val_mae = df_hist.loc[best_idx, "mae_mean"]

    # Nama file png mengikuti nama folder run
    run_name = history_path.parent.name
    save_path = save_dir / f"{run_name}_train_vs_val_mae.png"

    # Plot
    plt.figure(figsize=(10, 6))
    plt.plot(df_hist["epoch"], df_hist["train_mae"], marker="o", label="Train MAE")
    plt.plot(df_hist["epoch"], df_hist["mae_mean"], marker="o", label="Validation MAE")
    plt.scatter(best_epoch, best_val_mae, s=100, label=f"Best Val MAE (Epoch {best_epoch})")

    plt.title(f"Training vs Validation MAE\n{run_name}")
    plt.xlabel("Epoch")
    plt.ylabel("MAE")
    plt.xticks(df_hist["epoch"])
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()

    plt.savefig(save_path, dpi=200, bbox_inches="tight")
    plt.close()

    saved_files.append(save_path)
    print(f"[SAVED] {save_path}")

print(f"\nTotal grafik tersimpan: {len(saved_files)}")
```
