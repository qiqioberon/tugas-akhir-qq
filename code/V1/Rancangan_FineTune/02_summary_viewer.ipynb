# 02 — Quick Summary Viewer

Buka hasil tuning & report tanpa training ulang.


```python
from pathlib import Path
import json, pandas as pd

ROOT = Path('/workspace/ta_finetune')
OUT_ROOT = ROOT / 'outputs' / 'finetune_strict_wavlm_lora'
print("OUT_ROOT:", OUT_ROOT)

# list runs
runs = sorted([p for p in OUT_ROOT.iterdir() if p.is_dir()])
print("runs:", len(runs))
for r in runs[:10]:
    print("-", r.name)
```

```python
# pilih final run terbaru yang punya report.json
cands = []
for r in OUT_ROOT.iterdir():
    if (r / "report.json").exists():
        cands.append(r)
cands = sorted(cands, key=lambda p: p.stat().st_mtime, reverse=True)
assert cands, "Belum ada report.json. Jalankan notebook 01 dulu."
final_dir = cands[0]
print("Using:", final_dir)

report = json.loads((final_dir / "report.json").read_text())
print("BEST:", report["best"])
print("TEST:", report["test"])
```

```python
# tampilkan table hasil phase1/phase2
import pandas as pd
p1 = pd.DataFrame(report["phase1"]).sort_values("best_S", ascending=False)
p2 = pd.DataFrame(report["phase2"]).sort_values("best_S", ascending=False)
print("PHASE1:")
display(p1)
print("PHASE2:")
display(p2)
```
